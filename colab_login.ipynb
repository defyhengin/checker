Строка сессии — это полный доступ к аккаунту. Скопируйте её только в Render (TG_SESSION) и больше никуда.

```
from getpass import getpass
from telethon import TelegramClient
from telethon.sessions import StringSession

api_id = int(input("api_id (цифры с my.telegram.org): ").strip())
api_hash = getpass("api_hash: ").strip()

client = TelegramClient(StringSession(), api_id, api_hash,
                        device_model="tg-readonly-mcp", app_version="1.0")
await client.start(
    phone=lambda: input("Телефон в формате +7...: ").strip(),
    code_callback=lambda: input("Код из Telegram: ").strip(),
    password=lambda: getpass("Пароль 2FA (если включён): "),
)
me = await client.get_me()
session = client.session.save()
await client.disconnect()

print(f"Вход выполнен: {me.first_name}.")
print("Скопируйте строку ниже ЦЕЛИКОМ в TG_SESSION на Render:\n")
print(session)
print("\nПотом: меню ⋮ → Изменить → Очистить все результаты, и закройте блокнот, не сохраняя.")
```
